```python
async def search_channel(channel_name, query, limit=50):
    await client.start()

    channel = await client.get_entity(channel_name)

    results = []

    async for msg in client.iter_messages(
        channel,
        search=query,
        limit=limit
    ):
        results.append({
            "date": msg.date,
            "text": msg.text,
            "views": getattr(msg, "views", None)
        })

    return results
```

```python
import pandas as pd
import re

def extract_urls(text):
    if not text:
        return []
    return re.findall(r'https?://\S+|t\.me/\S+', text)

async def collect_channel_messages(channel_name, search=None, limit=100):
    await client.start()
    channel = await client.get_entity(channel_name)

    rows = []

    async for msg in client.iter_messages(
        channel,
        search=search,
        limit=limit
    ):
        text = msg.text or ""

        rows.append({
            "channel": channel_name,
            "message_id": msg.id,
            "date": msg.date,
            "text": text,
            "views": getattr(msg, "views", None),
            "forwards": getattr(msg, "forwards", None),
            "reply_to_msg_id": getattr(msg, "reply_to_msg_id", None),
            "has_media": msg.media is not None,
            "urls": extract_urls(text),
            "message_link": f"https://t.me/{channel_name}/{msg.id}"
        })

    return pd.DataFrame(rows)
```

```python
# ============================================================
# STEP 1: Clean and prepare Telegram messages
# ============================================================

import pandas as pd
import re

# Make a copy so we do not accidentally damage the original df
messages = df.copy()

# Ensure the date column is treated as datetime
messages["date"] = pd.to_datetime(messages["date"], errors="coerce")

# Remove rows with empty or missing text
messages["text"] = messages["text"].fillna("").astype(str)
messages = messages[messages["text"].str.strip() != ""]

# Remove duplicate messages, if any
messages = messages.drop_duplicates(subset=["message_id", "text"])

# Basic text cleaning for analysis
def clean_text(text):
    """
    Cleans Telegram message text while keeping useful OSINT signals.
    We remove excessive whitespace but keep hashtags, URLs, names, etc.
    """
    text = re.sub(r"\s+", " ", text)
    return text.strip()

messages["clean_text"] = messages["text"].apply(clean_text)

# Add simple helper fields
messages["text_length"] = messages["clean_text"].str.len()
messages["word_count"] = messages["clean_text"].str.split().str.len()
messages["date_only"] = messages["date"].dt.date
messages["hour"] = messages["date"].dt.hour

print("Original rows:", len(df))
print("Clean usable messages:", len(messages))
print("Date range:", messages["date"].min(), "to", messages["date"].max())

messages[["date", "clean_text", "views", "forwards", "message_link"]].head(5)
```

```python
# ============================================================
# STEP 2: Find most common words in the channel
# ============================================================

from collections import Counter
import re

# Common words that add little intelligence value
STOP_WORDS = {
    "the","a","an","and","or","but","of","to","in","on","for",
    "with","at","by","from","is","are","was","were","be","been",
    "has","have","had","that","this","it","its","as","after",
    "about","into","their","they","them","will","would","can",
    "could","should","may","might","you","your","we","our"
}

all_words = []

for text in messages["clean_text"]:

    # Extract words
    words = re.findall(r"\b[a-zA-Z]{3,}\b", text.lower())

    # Remove stop words
    words = [w for w in words if w not in STOP_WORDS]

    all_words.extend(words)

word_counts = Counter(all_words)

top_words = word_counts.most_common(50)

for word, count in top_words:
    print(f"{word:<20} {count}")
```

```python
# ============================================================
# STEP 3: Extract hashtags
# ============================================================

hashtags = []

for text in messages["clean_text"]:
    hashtags.extend(
        re.findall(r"#(\w+)", text)
    )

hashtag_counts = Counter(hashtags)

print("Top hashtags:\n")

for tag, count in hashtag_counts.most_common(30):
    print(f"#{tag:<25} {count}")
```

```python
# ============================================================
# STEP 4: Find referenced websites
# ============================================================

from urllib.parse import urlparse

domains = []

for urls in messages["urls"]:
    
    if not urls:
        continue

    for url in urls:
        try:
            domain = urlparse(url).netloc
            domain = domain.replace("www.", "")
            domains.append(domain)
        except:
            pass

domain_counts = Counter(domains)

print("Top referenced domains:\n")

for domain, count in domain_counts.most_common(20):
    print(f"{domain:<40} {count}")
```

```python
# ============================================================
# STEP 5A: Load spaCy
# ============================================================

import spacy

nlp = spacy.load("en_core_web_sm")

print("spaCy loaded")
```

```python
# ============================================================
# STEP 5B: Extract entities
# ============================================================

from collections import Counter

people = Counter()
orgs = Counter()
locations = Counter()

for text in messages["clean_text"]:

    doc = nlp(text)

    for ent in doc.ents:

        if ent.label_ == "PERSON":
            people[ent.text] += 1

        elif ent.label_ == "ORG":
            orgs[ent.text] += 1

        elif ent.label_ in ["GPE", "LOC"]:
            locations[ent.text] += 1
```

```python
print("\nTOP PEOPLE\n")

for name, count in people.most_common(20):
    print(f"{name:<30} {count}")
```

```python
print("\nTOP ORGANIZATIONS\n")

for org, count in orgs.most_common(20):
    print(f"{org:<40} {count}")
```

```python
print("\nTOP LOCATIONS\n")

for place, count in locations.most_common(20):
    print(f"{place:<30} {count}")
```

```python
# ============================================================
# STEP 6A: Load embedding model
# ============================================================

from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("Model loaded")
```

```python
# ============================================================
# STEP 6B: Convert messages into vectors
# ============================================================

texts = messages["clean_text"].tolist()

embeddings = model.encode(
    texts,
    show_progress_bar=True,
    convert_to_numpy=True
)

print("Embedding shape:", embeddings.shape)
```

```python
# ============================================================
# STEP 7A: Cluster messages
# ============================================================

from sklearn.cluster import KMeans

NUM_CLUSTERS = 8

kmeans = KMeans(
    n_clusters=NUM_CLUSTERS,
    random_state=42,
    n_init=10
)

messages["cluster"] = kmeans.fit_predict(
    embeddings
)

print(
    messages["cluster"].value_counts()
)
```

```python
# ============================================================
# STEP 8A: Review cluster examples
# ============================================================

for cluster_id in sorted(messages["cluster"].unique()):

    print("\n" + "=" * 80)
    print(f"CLUSTER {cluster_id}")
    print("=" * 80)

    sample = messages[
        messages["cluster"] == cluster_id
    ]["clean_text"].head(5)

    for msg in sample:
        print("\n•", msg[:300])
```

```python
from collections import Counter
import re

STOP_WORDS = {
    "the","and","for","that","with","this",
    "from","have","has","been","will","into",
    "after","their","about","they","were"
}

for cluster_id in sorted(messages["cluster"].unique()):

    cluster_messages = messages[
        messages["cluster"] == cluster_id
    ]

    words = []

    for text in cluster_messages["clean_text"]:

        tokens = re.findall(
            r"\b[a-zA-Z]{3,}\b",
            text.lower()
        )

        tokens = [
            t for t in tokens
            if t not in STOP_WORDS
        ]

        words.extend(tokens)

    common_words = Counter(words).most_common(10)

    print("\n")
    print("=" * 60)
    print(f"NARRATIVE CLUSTER {cluster_id}")
    print("=" * 60)
    print(common_words)
```

```python
# ============================================================
# STEP 10: Build a Narrative Intelligence Report
# ============================================================

from collections import Counter
import re
import pandas as pd

# Better stop words for narrative keyword extraction
STOP_WORDS = {
    "the","and","for","that","with","this","from","have","has","been",
    "will","into","after","their","about","they","were","are","was",
    "you","your","but","not","all","can","more","said","says","over",
    "new","now","out","his","her","its","who","why","how","what"
}

def get_top_keywords(texts, top_n=8):
    """
    Extracts common meaningful keywords from a group of messages.
    """
    words = []

    for text in texts:
        tokens = re.findall(r"\b[a-zA-Z]{3,}\b", text.lower())
        tokens = [t for t in tokens if t not in STOP_WORDS]
        words.extend(tokens)

    return Counter(words).most_common(top_n)


narrative_reports = []

for cluster_id in sorted(messages["cluster"].unique()):

    cluster_df = messages[messages["cluster"] == cluster_id].copy()

    # Basic size and time metrics
    message_count = len(cluster_df)
    first_seen = cluster_df["date"].min()
    last_seen = cluster_df["date"].max()

    # Engagement metrics
    total_views = cluster_df["views"].fillna(0).sum()
    avg_views = cluster_df["views"].fillna(0).mean()
    total_forwards = cluster_df["forwards"].fillna(0).sum()

    # Keywords give us a rough narrative label
    keywords = get_top_keywords(cluster_df["clean_text"].tolist(), top_n=8)

    # Pick representative messages:
    # highest-viewed messages are often the most useful analyst examples
    representative = (
        cluster_df
        .sort_values("views", ascending=False)
        [["date", "views", "forwards", "clean_text", "message_link"]]
        .head(3)
    )

    narrative_reports.append({
        "cluster": cluster_id,
        "message_count": message_count,
        "first_seen": first_seen,
        "last_seen": last_seen,
        "total_views": total_views,
        "avg_views": round(avg_views, 2),
        "total_forwards": total_forwards,
        "keywords": keywords,
        "representative_messages": representative
    })


# Print the report in a readable analyst format
for report in narrative_reports:

    print("\n" + "=" * 90)
    print(f"NARRATIVE CLUSTER {report['cluster']}")
    print("=" * 90)

    print("Message count:", report["message_count"])
    print("First seen:", report["first_seen"])
    print("Last seen:", report["last_seen"])
    print("Total views:", report["total_views"])
    print("Average views:", report["avg_views"])
    print("Total forwards:", report["total_forwards"])

    print("\nTop keywords:")
    for word, count in report["keywords"]:
        print(f"  {word:<20} {count}")

    print("\nRepresentative messages:")
    for _, row in report["representative_messages"].iterrows():
        print("-" * 70)
        print("Date:", row["date"])
        print("Views:", row["views"])
        print("Forwards:", row["forwards"])
        print("Link:", row["message_link"])
        print("Text:", row["clean_text"][:500])
```

```python
# ============================================================
# STEP 11: Narrative Importance Ranking
# ============================================================

import pandas as pd

summary_rows = []

for report in narrative_reports:

    # Simple narrative score
    score = (
        report["message_count"] * 1
        + report["avg_views"] * 0.01
        + report["total_forwards"] * 5
    )

    summary_rows.append({
        "cluster": report["cluster"],
        "messages": report["message_count"],
        "avg_views": report["avg_views"],
        "total_forwards": report["total_forwards"],
        "score": round(score, 2),
        "keywords": ", ".join(
            [w for w, c in report["keywords"][:5]]
        )
    })

narrative_summary = pd.DataFrame(summary_rows)

narrative_summary = narrative_summary.sort_values(
    "score",
    ascending=False
)

narrative_summary
```
